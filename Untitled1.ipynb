```python
"""
================================================================================
  LINEAR REGRESSION — INDUSTRIAL STANDARD PIPELINE
  Dataset  : California Housing (sklearn.datasets)
  Author   : ML Pipeline v2.4.1
  Python   : 3.9+
  Requires : pip install scikit-learn pandas numpy matplotlib seaborn joblib
================================================================================
"""

# ── 0. IMPORTS ────────────────────────────────────────────────────────────────
import warnings
warnings.filterwarnings('ignore')

import numpy  as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns

from sklearn.datasets        import fetch_california_housing
from sklearn.model_selection import (train_test_split, KFold,
                                     cross_val_score, GridSearchCV,
                                     learning_curve)
from sklearn.preprocessing   import StandardScaler
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.linear_model    import (LinearRegression, Ridge,
                                     Lasso, ElasticNet)
from sklearn.pipeline        import Pipeline
from sklearn.metrics         import (mean_squared_error,
                                     mean_absolute_error,
                                     r2_score)
from scipy                   import stats
import joblib

# ── Plot style ────────────────────────────────────────────────────────────────
plt.style.use('dark_background')
PALETTE = {
    'blue'   : '#388bfd',
    'green'  : '#3fb950',
    'orange' : '#d29922',
    'red'    : '#f85149',
    'purple' : '#bc8cff',
    'teal'   : '#39d353',
    'gray'   : '#8b949e',
    'bg'     : '#0d1117',
    'bg2'    : '#161b22',
}
plt.rcParams.update({
    'figure.facecolor'  : PALETTE['bg'],
    'axes.facecolor'    : PALETTE['bg2'],
    'axes.edgecolor'    : '#30363d',
    'axes.labelcolor'   : PALETTE['gray'],
    'xtick.color'       : PALETTE['gray'],
    'ytick.color'       : PALETTE['gray'],
    'grid.color'        : '#21262d',
    'grid.linewidth'    : 0.5,
    'text.color'        : '#e6edf3',
    'font.family'       : 'monospace',
    'font.size'         : 10,
})

SEED = 42
np.random.seed(SEED)


# ══════════════════════════════════════════════════════════════════════════════
#  STEP 1 — DATA INGESTION & VALIDATION
# ══════════════════════════════════════════════════════════════════════════════

print("=" * 70)
print("  STEP 1 — DATA INGESTION & VALIDATION")
print("=" * 70)

housing = fetch_california_housing(as_frame=True)
df      = housing.frame

# ── Schema validation ────────────────────────────────────────────────────────
assert df.isnull().sum().sum() == 0, "Missing values detected — aborting"
assert df.shape == (20640, 9),       "Unexpected dataset shape"
assert 'MedHouseVal' in df.columns,  "Target column missing"

print(f"\n  Samples   : {df.shape[0]:,}")
print(f"  Features  : {df.shape[1] - 1}")
print(f"  Target    : MedHouseVal (Median House Value, $100k)")
print(f"  Missing   : {df.isnull().sum().sum()}")
print("\n  Feature dtypes:")
print(df.dtypes.to_string())
print("\n  Statistical summary:")
print(df.describe().round(3).to_string())


# ══════════════════════════════════════════════════════════════════════════════
#  STEP 2 — EXPLORATORY DATA ANALYSIS
# ══════════════════════════════════════════════════════════════════════════════

print("\n" + "=" * 70)
print("  STEP 2 — EXPLORATORY DATA ANALYSIS")
print("=" * 70)

# ── 2A  Correlation with target ───────────────────────────────────────────────
corr_with_target = df.corr()['MedHouseVal'].drop('MedHouseVal').sort_values()
print("\n  Pearson correlation with MedHouseVal:")
print(corr_with_target.round(3).to_string())

# ── 2B  Skewness report ────────────────────────────────────────────────────────
skew_df = pd.DataFrame({
    'skewness' : df.skew(),
    'kurtosis' : df.kurt()
}).round(3)
print("\n  Skewness & Kurtosis:")
print(skew_df.to_string())

# ── 2C  Outlier detection (IQR) ───────────────────────────────────────────────
def iqr_outlier_count(series):
    Q1, Q3 = series.quantile(0.25), series.quantile(0.75)
    IQR    = Q3 - Q1
    mask   = (series < Q1 - 1.5 * IQR) | (series > Q3 + 1.5 * IQR)
    return mask.sum()

outlier_counts = df.apply(iqr_outlier_count)
print("\n  IQR outlier counts:")
print(outlier_counts.to_string())

# ── 2D  EDA plots ─────────────────────────────────────────────────────────────
fig = plt.figure(figsize=(16, 10))
fig.suptitle('EDA — California Housing Dataset', fontsize=14,
             fontweight='bold', color='#e6edf3', y=1.01)
gs  = gridspec.GridSpec(2, 3, figure=fig, hspace=0.45, wspace=0.35)

# Target distribution
ax1 = fig.add_subplot(gs[0, 0])
ax1.hist(df['MedHouseVal'], bins=50,
         color=PALETTE['blue'], alpha=0.7, edgecolor='none')
ax1.set_title('Target Distribution', fontsize=10)
ax1.set_xlabel('MedHouseVal ($100k)')
ax1.set_ylabel('Count')
ax1.grid(True, alpha=0.3)

# MedInc vs target scatter
ax2 = fig.add_subplot(gs[0, 1])
ax2.scatter(df['MedInc'], df['MedHouseVal'],
            alpha=0.05, s=5, color=PALETTE['green'])
ax2.set_title('MedInc vs Target  (r=0.688)', fontsize=10)
ax2.set_xlabel('MedInc')
ax2.set_ylabel('MedHouseVal')
ax2.grid(True, alpha=0.3)

# Correlation bar chart
ax3 = fig.add_subplot(gs[0, 2])
colors = [PALETTE['green'] if v > 0 else PALETTE['red']
          for v in corr_with_target.values]
ax3.barh(corr_with_target.index, corr_with_target.values,
         color=colors, alpha=0.8)
ax3.axvline(0, color='white', linewidth=0.5)
ax3.set_title('Correlation with Target', fontsize=10)
ax3.grid(True, alpha=0.3, axis='x')

# Heatmap
ax4 = fig.add_subplot(gs[1, :])
corr_matrix = df.corr()
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix, ax=ax4, annot=True, fmt='.2f',
            cmap='RdYlGn', mask=mask, linewidths=0.3,
            annot_kws={'size': 8}, center=0,
            cbar_kws={'shrink': 0.8})
ax4.set_title('Correlation Heatmap', fontsize=10)

plt.tight_layout()
plt.savefig('eda_plots.png', dpi=150, bbox_inches='tight',
            facecolor=PALETTE['bg'])
plt.show()
print("\n  [Saved] eda_plots.png")


# ══════════════════════════════════════════════════════════════════════════════
#  STEP 3 — FEATURE ENGINEERING
# ══════════════════════════════════════════════════════════════════════════════

print("\n" + "=" * 70)
print("  STEP 3 — FEATURE ENGINEERING")
print("=" * 70)

df_fe = df.copy()

# ── Domain-driven features ────────────────────────────────────────────────────
df_fe['RoomsPerPerson']  = df_fe['AveRooms']    / df_fe['AveOccup']
df_fe['BedroomRatio']    = df_fe['AveBedrms']   / df_fe['AveRooms']
df_fe['PopDensity']      = df_fe['Population']  / df_fe['AveOccup']
df_fe['IncomePerRoom']   = df_fe['MedInc']      / df_fe['AveRooms']

print("  New features created:")
new_feats = ['RoomsPerPerson', 'BedroomRatio', 'PopDensity', 'IncomePerRoom']
print(df_fe[new_feats].describe().round(3).to_string())

# ── Clip extreme outliers at 99th percentile ─────────────────────────────────
clip_cols = ['AveRooms', 'AveBedrms', 'Population', 'AveOccup']
for col in clip_cols:
    upper = df_fe[col].quantile(0.99)
    n_clipped = (df_fe[col] > upper).sum()
    df_fe[col] = df_fe[col].clip(upper=upper)
    print(f"  Clipped {col}: {n_clipped} values above {upper:.2f}")

# ── Check new correlations ────────────────────────────────────────────────────
new_corrs = df_fe[new_feats + ['MedHouseVal']].corr()['MedHouseVal'].drop('MedHouseVal')
print("\n  New feature correlations with target:")
print(new_corrs.round(3).to_string())

# ── Train / Test split ────────────────────────────────────────────────────────
X = df_fe.drop(columns=['MedHouseVal'])
y = df_fe['MedHouseVal']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=SEED
)
print(f"\n  Train size : {X_train.shape}")
print(f"  Test size  : {X_test.shape}")


# ══════════════════════════════════════════════════════════════════════════════
#  STEP 4 — MODEL TRAINING WITH PIPELINE
# ══════════════════════════════════════════════════════════════════════════════

print("\n" + "=" * 70)
print("  STEP 4 — MODEL TRAINING (Pipeline)")
print("=" * 70)

# ── Build Pipeline: scaler → feature selection → model ───────────────────────
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('select', SelectKBest(f_regression, k=10)),
    ('model',  LinearRegression())
])

pipeline.fit(X_train, y_train)
y_pred = pipeline.predict(X_test)

print("\n  Pipeline steps:")
for name, step in pipeline.steps:
    print(f"    [{name}] {step.__class__.__name__}")

# ── Metrics ───────────────────────────────────────────────────────────────────
def print_metrics(name, y_true, y_pred):
    r2   = r2_score(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae  = mean_absolute_error(y_true, y_pred)
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    print(f"\n  {'─'*50}")
    print(f"  Model  : {name}")
    print(f"  R²     : {r2:.4f}")
    print(f"  RMSE   : {rmse:.4f}  (${rmse*100:.0f}k)")
    print(f"  MAE    : {mae:.4f}  (${mae*100:.0f}k)")
    print(f"  MAPE   : {mape:.2f}%")
    print(f"  {'─'*50}")
    return {'R2': r2, 'RMSE': rmse, 'MAE': mae, 'MAPE': mape}

lr_metrics = print_metrics("Linear Regression (OLS)", y_test, y_pred)

# ── Cross-Validation ──────────────────────────────────────────────────────────
print("\n  Running 10-Fold Cross-Validation...")
kf = KFold(n_splits=10, shuffle=True, random_state=SEED)
cv_r2   = cross_val_score(pipeline, X, y, cv=kf, scoring='r2',
                          n_jobs=-1)
cv_rmse = np.sqrt(-cross_val_score(pipeline, X, y, cv=kf,
                                   scoring='neg_mean_squared_error',
                                   n_jobs=-1))
print(f"\n  CV R²  : {cv_r2.mean():.4f} ± {cv_r2.std():.4f}")
print(f"  CV RMSE: {cv_rmse.mean():.4f} ± {cv_rmse.std():.4f}")
print(f"  Fold R²: {np.round(cv_r2, 3)}")

# ── Learning curve ────────────────────────────────────────────────────────────
print("\n  Computing learning curve...")
train_sizes, train_scores, val_scores = learning_curve(
    pipeline, X, y,
    train_sizes=np.linspace(0.1, 1.0, 10),
    cv=5, scoring='r2', n_jobs=-1
)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(train_sizes, train_scores.mean(axis=1),
        color=PALETTE['blue'], label='Train R²', linewidth=2)
ax.fill_between(train_sizes,
                train_scores.mean(axis=1) - train_scores.std(axis=1),
                train_scores.mean(axis=1) + train_scores.std(axis=1),
                alpha=0.2, color=PALETTE['blue'])
ax.plot(train_sizes, val_scores.mean(axis=1),
        color=PALETTE['green'], label='Val R²', linewidth=2)
ax.fill_between(train_sizes,
                val_scores.mean(axis=1) - val_scores.std(axis=1),
                val_scores.mean(axis=1) + val_scores.std(axis=1),
                alpha=0.2, color=PALETTE['green'])
ax.set_title('Learning Curve — Linear Regression', fontsize=12,
             fontweight='bold')
ax.set_xlabel('Training Samples')
ax.set_ylabel('R² Score')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('learning_curve.png', dpi=150, bbox_inches='tight',
            facecolor=PALETTE['bg'])
plt.show()
print("  [Saved] learning_curve.png")


# ══════════════════════════════════════════════════════════════════════════════
#  STEP 5 — MODEL EVALUATION & RESIDUAL ANALYSIS
# ══════════════════════════════════════════════════════════════════════════════

print("\n" + "=" * 70)
print("  STEP 5 — MODEL EVALUATION & RESIDUAL ANALYSIS")
print("=" * 70)

residuals  = y_test - y_pred
std_resid  = (residuals - residuals.mean()) / residuals.std()

# ── Assumption tests ──────────────────────────────────────────────────────────
# Normality (Shapiro-Wilk on a sample)
sample_idx   = np.random.choice(len(residuals), size=min(5000, len(residuals)),
                                replace=False)
sw_stat, sw_p = stats.shapiro(residuals.values[sample_idx])
print(f"\n  Normality (Shapiro-Wilk) : stat={sw_stat:.4f}, p={sw_p:.4f}")
print(f"  → {'PASS: Residuals approx. normal' if sw_p > 0.05 else 'FAIL: Non-normal residuals (consider log transform)'}")

# Homoscedasticity (Breusch-Pagan via correlation proxy)
abs_resid = np.abs(residuals)
corr_bp, p_bp = stats.pearsonr(y_pred, abs_resid)
print(f"\n  Homoscedasticity (BP proxy) : corr={corr_bp:.4f}, p={p_bp:.4f}")
print(f"  → {'PASS: Constant variance' if p_bp > 0.05 else 'WARN: Heteroscedasticity detected'}")

# Durbin-Watson
dw = np.sum(np.diff(residuals.values)**2) / np.sum(residuals.values**2)
print(f"\n  Durbin-Watson statistic     : {dw:.4f}")
print(f"  → {'PASS: No autocorrelation' if 1.5 < dw < 2.5 else 'WARN: Possible autocorrelation'}")

# ── Feature coefficients ──────────────────────────────────────────────────────
selector   = pipeline.named_steps['select']
model_step = pipeline.named_steps['model']
selected_mask   = selector.get_support()
selected_feats  = X.columns[selected_mask]
coef_df = pd.DataFrame({
    'Feature'    : selected_feats,
    'Coefficient': model_step.coef_
}).sort_values('Coefficient', key=abs, ascending=False)

print("\n  Feature Coefficients (standardized):")
print(coef_df.to_string(index=False))

# ── Evaluation plots (2×3 grid) ───────────────────────────────────────────────
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle('Model Evaluation — Linear Regression', fontsize=14,
             fontweight='bold', color='#e6edf3')

# 1 — Actual vs Predicted
axes[0,0].scatter(y_test, y_pred, alpha=0.15, s=6,
                  color=PALETTE['green'])
lims = [min(y_test.min(), y_pred.min()),
        max(y_test.max(), y_pred.max())]
axes[0,0].plot(lims, lims, '--', color=PALETTE['red'],
               linewidth=1.5, label='Perfect fit')
axes[0,0].set_title('Actual vs Predicted')
axes[0,0].set_xlabel('Actual')
axes[0,0].set_ylabel('Predicted')
axes[0,0].legend(fontsize=8)
axes[0,0].grid(True, alpha=0.3)

# 2 — Residuals vs Predicted
axes[0,1].scatter(y_pred, residuals, alpha=0.15, s=6,
                  color=PALETTE['blue'])
axes[0,1].axhline(0, color=PALETTE['red'], linewidth=1.5, linestyle='--')
axes[0,1].set_title('Residuals vs Predicted')
axes[0,1].set_xlabel('Predicted')
axes[0,1].set_ylabel('Residuals')
axes[0,1].grid(True, alpha=0.3)

# 3 — Residual distribution
axes[0,2].hist(residuals, bins=60, color=PALETTE['purple'],
               alpha=0.75, edgecolor='none', density=True)
xmin, xmax = axes[0,2].get_xlim()
xp = np.linspace(xmin, xmax, 200)
axes[0,2].plot(xp, stats.norm.pdf(xp, residuals.mean(), residuals.std()),
               color=PALETTE['orange'], linewidth=2, label='Normal fit')
axes[0,2].set_title('Residual Distribution')
axes[0,2].set_xlabel('Residual')
axes[0,2].legend(fontsize=8)
axes[0,2].grid(True, alpha=0.3)

# 4 — Q-Q plot
(osm, osr), (slope, intercept, _) = stats.probplot(residuals)
axes[1,0].scatter(osm, osr, alpha=0.3, s=6, color=PALETTE['teal'])
line_x = np.array([osm.min(), osm.max()])
axes[1,0].plot(line_x, slope * line_x + intercept,
               color=PALETTE['red'], linewidth=1.5, linestyle='--')
axes[1,0].set_title('Q-Q Plot (Normality)')
axes[1,0].set_xlabel('Theoretical Quantiles')
axes[1,0].set_ylabel('Sample Quantiles')
axes[1,0].grid(True, alpha=0.3)

# 5 — Scale-location (sqrt|residuals| vs predicted)
axes[1,1].scatter(y_pred, np.sqrt(np.abs(std_resid)),
                  alpha=0.15, s=6, color=PALETTE['orange'])
axes[1,1].set_title('Scale-Location (Homoscedasticity)')
axes[1,1].set_xlabel('Predicted')
axes[1,1].set_ylabel('√|Standardized Residuals|')
axes[1,1].grid(True, alpha=0.3)

# 6 — Feature importance bar
top_n = coef_df.head(8)
bar_colors = [PALETTE['green'] if c > 0 else PALETTE['red']
              for c in top_n['Coefficient']]
axes[1,2].barh(top_n['Feature'], top_n['Coefficient'],
               color=bar_colors, alpha=0.85)
axes[1,2].axvline(0, color='white', linewidth=0.5)
axes[1,2].set_title('Feature Coefficients (Top 8)')
axes[1,2].set_xlabel('Coefficient Value')
axes[1,2].grid(True, alpha=0.3, axis='x')

plt.tight_layout()
plt.savefig('evaluation_plots.png', dpi=150, bbox_inches='tight',
            facecolor=PALETTE['bg'])
plt.show()
print("  [Saved] evaluation_plots.png")


# ══════════════════════════════════════════════════════════════════════════════
#  STEP 6 — REGULARIZATION (Ridge, Lasso, ElasticNet, GridSearchCV)
# ══════════════════════════════════════════════════════════════════════════════

print("\n" + "=" * 70)
print("  STEP 6 — REGULARIZATION & MODEL COMPARISON")
print("=" * 70)

# ── Define all models ─────────────────────────────────────────────────────────
models = {
    'Linear Regression' : LinearRegression(),
    'Ridge α=0.1'       : Ridge(alpha=0.1),
    'Ridge α=1.0'       : Ridge(alpha=1.0),
    'Ridge α=10'        : Ridge(alpha=10.0),
    'Lasso α=0.01'      : Lasso(alpha=0.01, max_iter=10000),
    'Lasso α=0.1'       : Lasso(alpha=0.1,  max_iter=10000),
    'ElasticNet α=0.05' : ElasticNet(alpha=0.05, l1_ratio=0.5, max_iter=10000),
}

results = {}
for name, mdl in models.items():
    pipe = Pipeline([
        ('scaler', StandardScaler()),
        ('model',  mdl)
    ])
    pipe.fit(X_train, y_train)
    preds = pipe.predict(X_test)
    results[name] = {
        'R2'  : round(r2_score(y_test, preds),               4),
        'RMSE': round(np.sqrt(mean_squared_error(y_test, preds)), 4),
        'MAE' : round(mean_absolute_error(y_test, preds),    4),
    }

results_df = pd.DataFrame(results).T
print("\n  Model Comparison:")
print(results_df.to_string())

# ── Grid Search: best Ridge alpha ─────────────────────────────────────────────
print("\n  Running GridSearchCV for Ridge alpha...")
ridge_pipe   = Pipeline([('scaler', StandardScaler()),
                         ('model',  Ridge())])
param_grid   = {'model__alpha': [0.001, 0.01, 0.1, 1, 10, 50, 100]}
gs = GridSearchCV(ridge_pipe, param_grid, cv=10,
                  scoring='r2', n_jobs=-1, verbose=0)
gs.fit(X_train, y_train)

print(f"\n  Best alpha       : {gs.best_params_['model__alpha']}")
print(f"  Best CV R²       : {gs.best_score_:.4f}")
print(f"  Test R²          : {r2_score(y_test, gs.predict(X_test)):.4f}")

# ── Regularization path (Lasso) ───────────────────────────────────────────────
print("\n  Computing Lasso regularization path...")
alphas     = np.logspace(-3, 1, 50)
lasso_coefs = []

for a in alphas:
    p = Pipeline([('scaler', StandardScaler()),
                  ('model',  Lasso(alpha=a, max_iter=10000))])
    p.fit(X_train, y_train)
    lasso_coefs.append(p.named_steps['model'].coef_)

lasso_coefs = np.array(lasso_coefs)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Regularization Analysis', fontsize=13,
             fontweight='bold', color='#e6edf3')

# Lasso path
for i in range(lasso_coefs.shape[1]):
    axes[0].plot(np.log10(alphas), lasso_coefs[:, i],
                 linewidth=1, alpha=0.8)
axes[0].axvline(np.log10(0.01), color=PALETTE['red'],
                linestyle='--', linewidth=1.5,
                label='Best α=0.01')
axes[0].set_title('Lasso Regularization Path')
axes[0].set_xlabel('log₁₀(Alpha)')
axes[0].set_ylabel('Coefficient Value')
axes[0].legend(fontsize=8)
axes[0].grid(True, alpha=0.3)

# Model R² comparison
r2_vals = [v['R2'] for v in results.values()]
bar_c   = [PALETTE['blue'] if 'Linear' in k
           else PALETTE['green'] if 'Ridge' in k
           else PALETTE['orange'] if 'Lasso' in k
           else PALETTE['purple']
           for k in results.keys()]
axes[1].barh(list(results.keys()), r2_vals,
             color=bar_c, alpha=0.8)
axes[1].axvline(0, color='white', linewidth=0.5)
axes[1].set_title('R² Score Comparison')
axes[1].set_xlabel('R² Score')
axes[1].set_xlim(0.55, 0.65)
axes[1].grid(True, alpha=0.3, axis='x')

plt.tight_layout()
plt.savefig('regularization_plots.png', dpi=150,
            bbox_inches='tight', facecolor=PALETTE['bg'])
plt.show()
print("  [Saved] regularization_plots.png")


# ══════════════════════════════════════════════════════════════════════════════
#  STEP 7 — MODEL SERIALIZATION & REGISTRY
# ══════════════════════════════════════════════════════════════════════════════

print("\n" + "=" * 70)
print("  STEP 7 — MODEL SERIALIZATION & REGISTRY")
print("=" * 70)

# ── Save full pipeline ────────────────────────────────────────────────────────
MODEL_PATH = 'lr_pipeline_v2.4.1.pkl'
joblib.dump(pipeline, MODEL_PATH, compress=3)
print(f"\n  [Saved] Pipeline → {MODEL_PATH}")

# ── Reload and verify ─────────────────────────────────────────────────────────
loaded_model = joblib.load(MODEL_PATH)
loaded_preds = loaded_model.predict(X_test)
assert np.allclose(y_pred, loaded_preds), "Loaded model predictions mismatch!"
print(f"  [Verified] Loaded model matches original predictions ✓")

# ── Inference example ────────────────────────────────────────────────────────
sample = X_test.iloc[:5]
predictions = loaded_model.predict(sample)
print(f"\n  Sample Predictions (first 5 test rows):")
print(f"  {'Predicted ($100k)':<22} {'Actual ($100k)':<18} {'Error'}")
print(f"  {'─'*55}")
for pred, actual in zip(predictions, y_test.iloc[:5]):
    err = pred - actual
    print(f"  {pred:>10.4f}           {actual:>10.4f}          {err:+.4f}")

# ── Model card summary ────────────────────────────────────────────────────────
print("\n" + "=" * 70)
print("  MODEL CARD — FINAL SUMMARY")
print("=" * 70)
print(f"""
  Model ID     : lr-california-v2.4.1
  Algorithm    : Linear Regression (OLS) + StandardScaler + SelectKBest
  Framework    : scikit-learn
  Dataset      : California Housing  (20,640 rows, 12 features)
  Train size   : {X_train.shape[0]:,}  |  Test size: {X_test.shape[0]:,}
  Random state : {SEED}

  ── Performance ──────────────────────────────────────────
  R²           : {lr_metrics['R2']:.4f}
  RMSE         : {lr_metrics['RMSE']:.4f}  (${lr_metrics['RMSE']*100:.0f}k error)
  MAE          : {lr_metrics['MAE']:.4f}  (${lr_metrics['MAE']*100:.0f}k error)
  CV R² (10-K) : {cv_r2.mean():.4f} ± {cv_r2.std():.4f}

  ── Assumptions ──────────────────────────────────────────
  Normality (SW)     : p = {sw_p:.4f}  {'✓' if sw_p > 0.05 else '⚠ slight deviation'}
  Homoscedasticity   : p = {p_bp:.4f}  {'✓' if p_bp > 0.05 else '⚠ heteroscedastic'}
  Durbin-Watson      : {dw:.4f}  {'✓ no autocorrelation' if 1.5 < dw < 2.5 else '⚠'}

  ── Artifacts ────────────────────────────────────────────
  Model pkl    : lr_pipeline_v2.4.1.pkl
  EDA plots    : eda_plots.png
  Eval plots   : evaluation_plots.png
  Reg. plots   : regularization_plots.png
  Learn curve  : learning_curve.png
""")
print("=" * 70)
print("  PIPELINE COMPLETE ✓")
print("=" * 70)
```
