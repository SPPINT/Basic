9.	What are the skewness values of numerical variables?

```python
{
    name: df.select_dtypes(include="number")
            .skew()
            .loc[lambda x: abs(x) > 1] # skew greater than 1
            .round(2)
    for name, df in files.items()
}
```

12.	Which features are most correlated with the target?
13.	Are there multicollinearity issues (correlation > 0.8)?

```python
corr = data.corr(numeric_only=True)['Salary'].drop('Salary').sort_values(ascending=False)

print("Most correlated features with target:")
print(corr)
corr1=corr[corr>0.8]
corr.head(1),corr1
```

```python
corr = diamond.corr(numeric_only=True)['price'].drop('price').sort_values(ascending=False)
print("High Correlated feature with target & correlation >0.8")
corr1=corr[corr>0.8]
corr1=corr[corr>0.8]
corr.head(1),corr1
```

```python
corr=housing.corr(numeric_only=True)['price'].drop('price').sort_values(ascending=False)
print("High Correlated feature with target & correlation >0.8")
corr1=corr[corr>0.8]
corr.head(1),corr1
```

```python
corr=car.corr(numeric_only=True)['Price'].drop('Price').sort_values(ascending=False)
print("High Correlated feature with target & correlation >0.8")
corr1=corr[corr>0.8]
corr.head(1),corr1
```
